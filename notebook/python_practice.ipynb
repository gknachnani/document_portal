```python
import traceback, sys

class AppError(Exception):
    pass

def demo(use_from):
    try:
        1 / 0                               # original error
    except Exception as e:
        if use_from:
            raise AppError("wrap with from") from e
        else:
            raise AppError("wrap NO from")

try:
    demo(use_from=False)                    # Change to True for the second run
except Exception as outer:
    traceback.print_exception(type(outer), outer, outer.__traceback__)
    print("\n__cause__   :", outer.__cause__)
    print("__context__ :", outer.__context__)
```

```python
import traceback, sys

class AppError(Exception):
    pass

def demo(use_from):
    try:
        1 / 0                               # original error
    except Exception as e:
        if use_from:
            raise AppError("wrap with from") from e
        else:
            raise AppError("wrap NO from")

try:
    demo(use_from=True)                    # Change to True for the second run
except Exception as outer:
    traceback.print_exception(type(outer), outer, outer.__traceback__)
    print("\n__cause__   :", outer.__cause__)
    print("__context__ :", outer.__context__)
```

```python
from datetime import datetime, timezone
import uuid

def make_session_id(session_id=None):
    return session_id or f"session_{datetime.now(timezone.utc).strftime('%Y%m%d_%H%M%S')}_{uuid.uuid4().hex[:8]}"

print(make_session_id("my_manual_id"))
# → my_manual_id   (user-supplied string wins)

print(make_session_id())
# → session_20250805_142310_b3e1f755   (auto-generated fallback)
```

```python
from pathlib import Path
import shutil, tempfile
from datetime import datetime, timezone
import uuid


# Fake session environment ---------------------------------------------
tmp = Path(tempfile.mkdtemp())
for days_ago in range(5):
    ts = (datetime.now(timezone.utc)
          .replace(hour=12, minute=0, second=0, microsecond=0)
          .strftime('%Y%m%d_%H%M%S'))
    sid = uuid.uuid4().hex[:8]
    folder = tmp / f"session_{ts}_{sid}"
    folder.mkdir()
    (folder / "dummy.txt").write_text("data")

# ------------------------------------------------------------------------
class Cleaner:
    def __init__(self, base_dir):
        self.base_dir = Path(base_dir)
        #from loguru import logger           # simple logger for demo
        #self.log = logger

    def clean_old_sessions(self, keep_latest:int = 3):
        try:
            session_folders = sorted([f for f in self.base_dir.iterdir()
                                    if f.is_dir()], reverse=True)
            for folder in session_folders[keep_latest:]:
                for file in folder.iterdir():
                    file.unlink()
                folder.rmdir()
#                self.log.info(f"Deleted {folder}")
                print(f"Deleted {folder}")
        except Exception as e:
#            self.log.error(f"Error: {e}")
            print(f"Error: {e}")

Cleaner(tmp).clean_old_sessions(keep_latest=2)
print(list(tmp.iterdir()))  # → shows only the two newest remain
shutil.rmtree(tmp)          # cleanup temp
```
