```python
def demo(a, b, c):
    print(a, b, c)

t = ("A", "B", "C")

demo(*t)   # Prints: A B C
demo(t)    # TypeError: demo() missing 2 required positional arguments: 'b' and 'c'
```

```python
import traceback, sys

class AppError(Exception):
    pass

def demo(use_from):
    try:
        1 / 0                               # original error
    except Exception as e:
        if use_from:
            raise AppError("wrap with from") from e
        else:
            raise AppError("wrap NO from")

try:
    demo(use_from=False)                    # Change to True for the second run
except Exception as outer:
    traceback.print_exception(type(outer), outer, outer.__traceback__)
    print("\n__cause__   :", outer.__cause__)
    print("__context__ :", outer.__context__)
```
